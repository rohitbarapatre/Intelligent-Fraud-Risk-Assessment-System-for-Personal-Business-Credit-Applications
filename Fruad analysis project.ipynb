# **Intelligent Fraud Risk Assessment System for Personal & Business Credit Applications**


# 1. Introduction

Fraud detection in the banking industry is one of the most critical operational functions because financial institutions handle sensitive customer information and high-value financial transactions daily. Credit card application fraud can lead to financial losses, reputational damage, compliance risks, and poor customer experience.

Banks process thousands of applications every day, making manual verification difficult and time-consuming. Fraud prevention teams therefore use a combination of credit bureau analysis, identity verification, behavioral analysis, and risk scoring models to identify suspicious applications.

This project is based on a simulated real-world fraud review workflow for both personal and business credit card applications. The objective of the project is to design a structured fraud risk assessment framework using SQL-driven analysis and business rules.



# 2. Problem Statement

Manual verification of credit card applications often creates operational inefficiencies because investigators must evaluate multiple risk indicators before making decisions. Inconsistent decision-making may increase the chances of fraudulent approvals or unnecessary rejection of genuine customers.

The organization therefore requires a centralized fraud risk assessment system capable of:
- Identifying suspicious applications
- Reducing false approvals
- Improving review consistency
- Supporting both personal and business applications
- Automating risk prioritization

# 7. Fraud Detection Methodology

## Red Flags
- More than 5 credit enquiries
- Utilization below 10%
- Utilization above 60%
- Utilization above 80% requiring escalation

## Fraud Characteristics
- Name mismatch
- SSN mismatch
- DOB mismatch
- Address mismatch
- Email mismatch

## Email Risk
- Low Risk
- Medium Risk
- High Risk

## Document Verification
If records cannot be verified internally or through bureau reports:
- SSN Card
- State ID
- Passport
- Business registration documents
- Tax documents
are requested for validation.



# 8. Risk Scoring Model

| Condition | Risk Points |
|---|---|
| Enquiries > 5 | +2 |
| Utilization > 60% | +2 |
| Email Risk = High | +3 |
| Name Mismatch | +2 |
| SSN/ITIN Mismatch | +3 |
| SSN/ITIN Alert | +4 |

## Decision Logic

| Score | Decision |
|---|---|
| 0-3 | Approve |
| 4-6 | Review |
| 7+ | Reject / Escalate |


# 9. SQL Implementation

9.1 Total Applications Query

```python
SELECT COUNT(*) AS Total_Applications FROM applications;
```

9.2 Personal vs Business Applications

```python
SELECT Application_Type, COUNT(*) FROM applications GROUP BY Application_Type;
```

9.3 Approval Rate Analysis

```python
SELECT Final_Decision, COUNT(*) FROM applications GROUP BY Final_Decision;
```

9.4 High Utilization Detection

```python
SELECT * FROM applications WHERE Utilization_Percent > 80;
```

9.5 High Email Risk Customers

```python
SELECT * FROM applications WHERE Email_Risk='High';
```

9.6 SSN/ITIN Alert Cases

```python
SELECT COUNT(*) FROM applications WHERE SSN_ITIN_Alert='Yes';
```

9.7 Average Utilization Analysis

```python
SELECT AVG(Utilization_Percent) FROM applications;
```

9.8 Document Verification Cases

```python
SELECT * FROM applications WHERE Docs_Required='Yes';
```

9.9 Name Mismatch Detection

```python
SELECT * FROM applications WHERE Name_Match='No';
```

9.10 Address Mismatch Analysis

```python
SELECT * FROM applications WHERE Address_Match='No';
```

9.11 Email Risk Distribution

```python
SELECT Email_Risk, COUNT(*) FROM applications GROUP BY Email_Risk;
```

9.12 Business vs Personal Utilization

```python
SELECT Application_Type, AVG(Utilization_Percent) FROM applications GROUP BY Application_Type;
```

9.13 Escalated Applications

```python
SELECT * FROM applications WHERE Final_Decision='Escalate';
```

9.14 Manual Review Cases

```python
SELECT * FROM applications WHERE Final_Decision='Review';
```

9.15 High Credit Enquiry Customers

```python
SELECT COUNT(*) FROM applications WHERE CBR_Enquiries_Last_6M > 5;
```

9.16 SSN/ITIN Mismatch Detection

```python
SELECT * FROM applications WHERE SSN_ITIN_Match='No';
```

9.17 Document Verification Rate

```python
SELECT ROUND(SUM(CASE WHEN Docs_Required='Yes' THEN 1 ELSE 0 END)*100.0/COUNT(*),2) AS Verification_Rate FROM applications;
```

9.18 Combined High-Risk Applications

```python
SELECT * FROM applications WHERE Email_Risk='High' AND Utilization_Percent > 80;
```

9.19 Business Applications Requiring Documents

```python
SELECT * FROM applications WHERE Application_Type='Business' AND Docs_Required='Yes';
```

9.20 Risk Score Calculation

```python
SELECT Application_ID,
(CASE WHEN CBR_Enquiries_Last_6M > 5 THEN 2 ELSE 0 END +
CASE WHEN Utilization_Percent > 60 THEN 2 ELSE 0 END +
CASE WHEN Email_Risk='High' THEN 3 ELSE 0 END +
CASE WHEN SSN_ITIN_Alert='Yes' THEN 4 ELSE 0 END) AS Risk_Score
FROM applications;
```


# 10. Analysis & Insights

- Applications with utilization above 80% showed higher fraud probability.
- SSN/ITIN alerts strongly correlated with rejection and escalation.
- High-risk email categories were commonly associated with suspicious applications.
- Business applications required more document verification compared to personal applications.
- Multiple credit enquiries within a short period increased risk exposure.
